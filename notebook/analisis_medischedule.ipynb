```python
# MediSchedule - Análisis de Tiempos de Espera y Causas de Inasistencia
# Programación con Estructuras de Datos - Semana 3
# Universidad: UNAN-Managua

from statistics import mean, median


# 1. Datos de los 10 pacientes

tiempos_espera = [45, 28, 52, 35, 41, 22, 38, 60, 25, 33]
asistencias = ["No", "Sí", "No", "Sí", "No", "Sí", "Sí", "No", "Sí", "No"]
causas = [
    "Olvido",
    None,
    "Dificultad para comunicarse",
    None,
    "Olvido",
    None,
    None,
    "Olvido",
    None,
    "Dificultad para comunicarse"
]


# 2. Función: Validar datos

def validar_datos(tiempos):
    for t in tiempos:
        if not isinstance(t, (int, float)) or t <= 0:
            raise ValueError("Se encontró un tiempo de espera inválido.")
    return True


# 3. Función: Calcular estadísticas

def calcular_estadisticas(tiempos):
    media = mean(tiempos)
    mediana = median(tiempos)
    
    mas_de_30 = 0
    for t in tiempos:
        if t > 30:
            mas_de_30 += 1
    
    porcentaje = (mas_de_30 / len(tiempos)) * 100
    return media, mediana, mas_de_30, porcentaje


# 4. Función: Analizar causas de inasistencia

def analizar_causas(asistencias, causas):
    olvido = 0
    comunicacion = 0
    otras = 0
    
    for i in range(len(asistencias)):
        if asistencias[i] == "No":
            if causas[i] == "Olvido":
                olvido += 1
            elif causas[i] == "Dificultad para comunicarse":
                comunicacion += 1
            else:
                otras += 1
                
    return olvido, comunicacion, otras


# 5. Función: Generar recomendaciones

def generar_recomendaciones(porcentaje, olvido, comunicacion):
    print("\n----- RECOMENDACIONES -----")
    
    if porcentaje > 50:
        print("• Implementar recordatorios automáticos y mejorar la gestión de agendas.")
    
    if olvido >= 2:
        print("• Enviar recordatorios por SMS o WhatsApp 24 horas antes de la cita.")
    
    if comunicacion >= 1:
        print("• Permitir cancelación y reprogramación fácil desde la aplicación.")
    
    print("• Digitalizar el control de citas para reducir tiempos de espera.")


# 6. Función principal

def principal():
    try:
        print("=== ANÁLISIS MEDISCHEDULE - TIEMPOS DE ESPERA ===\n")
        
        # Validar
        validar_datos(tiempos_espera)
        print("✓ Datos validados correctamente")
        
        # Estadísticas
        media, mediana, mas_de_30, porcentaje = calcular_estadisticas(tiempos_espera)
        
        print(f"\nPromedio de espera: {media:.1f} minutos")
        print(f"Mediana de espera: {mediana:.1f} minutos")
        print(f"Pacientes que esperaron más de 30 min: {mas_de_30} de {len(tiempos_espera)}")
        print(f"Porcentaje que esperó más de 30 min: {porcentaje:.0f}%")
        
        # Causas
        olvido, comunicacion, otras = analizar_causas(asistencias, causas)
        
        print("\n----- CAUSAS DE INASISTENCIA -----")
        print(f"Por olvido: {olvido}")
        print(f"Por dificultad de comunicación: {comunicacion}")
        print(f"Otras causas: {otras}")
        
        # Recomendaciones
        generar_recomendaciones(porcentaje, olvido, comunicacion)
        
        print("\n=== Análisis finalizado ===")
        
    except ValueError as e:
        print(f"Error de validación: {e}")
    except Exception as e:
        print(f"Ocurrió un error inesperado: {e}")


principal()
```
